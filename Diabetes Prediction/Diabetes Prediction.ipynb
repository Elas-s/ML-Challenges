```python
#separating data and labels

X=diabetes_dataset.drop(columns='Outcome', axis=1) #axis should be specified
Y= diabetes_dataset['Outcome']
```

# Data Standaralization

```python
scaler= StandardScaler()
```

```python
scaler.fit(X)
```

```python
standarized_data = scaler.transform(X)
```

```python
print(standarized_data)
```

```python
X = standarized_data
Y = diabetes_dataset['Outcome']
```

```python
print(X,Y)
```

Train Test Split

```python
X_train, X_test, Y_train, Y_test =  train_test_split(X,Y, test_size = 0.2, stratify = Y, random_state =2)
```

```python
print(X.shape, Y.shape, X_test.shape)
```

# training the model

```python
classifier = svm.SVC(kernel='linear')
```

```python
#trainging the support vector Machine Classifieer

classifier.fit(X_train, Y_train)
```

# Model Evaluation

1- Accuracy Score

```python
# accuracy score on the training data

X_train_prediction = classifier.predict(X_train)
training_data_accuracy = accuracy_score (X_train_prediction, Y_train)
```

```python
print('Accuracy score of the training data: ', training_data_accuracy)
```

```python
# accuracy score on the testing data

X_test_prediction = classifier.predict(X_test)
test_data_accuracy = accuracy_score (X_test_prediction, Y_test)
```

```python
print('Accuracy score of the test data: ', test_data_accuracy)
```

---

### Why training accuracy is the same

Both models were trained on the **same scaled data**, with the **same split**, and this dataset is relatively simple and low-dimensional (8 features). Both models converge to a very similar decision boundary on the training set, they're both linear classifiers, so on clean, standardized data they can land on the same training accuracy by coincidence.

### Why test accuracy differs

The test set is **unseen data**:

- SVM optimizes for the **maximum margin**, it finds the boundary that is as far as possible from both classes. This makes it **generalize better** to new points.
- LR optimizes for **log-likelihood on the training set**, it fits the training data well, but the boundary isn't necessarily the most "cautious" one.
- SVM only cares about **support vectors** (the hardest points near the boundary). LR is influenced by all 614 training points, including easy ones far from the boundary, which can slightly skew the boundary.

In other words: **SVM's margin maximization is specifically designed to generalize**, which is why it edges ahead on unseen data.
